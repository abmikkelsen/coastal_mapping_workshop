### option B: use a helper function and add only the corners of a polygon 
(e.g. get coordinate from google maps or similar)

```python
## helper function to create a polygon from opposite corners
def bbox_to_polygon(corner1, corner2):
    """corner1, corner2: [lon, lat] of opposite corners, in any order."""
    
    ### NOTE - written for googe maps format with lat first. check which is your first coordinate
    lat1, lon1, = corner1
    lat2, lon2, = corner2

    min_lon, max_lon = sorted((lon1, lon2))
    min_lat, max_lat = sorted((lat1, lat2))

    return [
        [min_lon, min_lat],  # bottom-left
        [max_lon, min_lat],  # bottom-right
        [max_lon, max_lat],  # top-right
        [min_lon, max_lat],  # top-left
        [min_lon, min_lat],  # close the ring
    ]
```
